Control Structures in R: The for loop

```r
for(i in 1:5) {print(i^2)}
```

```r
for(i in c(2,4,6,7)) {print(i^2)}
```

```r
x = c(2,4,6,8,10,12)
excount = function(x) {
  count = 0
  for(xval in x) {
    if(xval/2 > 3)
    count = count + 1
  }
  print(count)
}
excount(x)
```

Nested looping with for loop

```r
child = c("child1", "child2", "child3")
sweet = c("sweet1", "sweet2", "sweet3")
for(x in child) {
  for(y in sweet) {
    print(paste(x,y))
  }
}
```

The break command

```r
drink = c("coffee" , "lemonade", "tea", "juice")
drink
```

```r
for(x in drink) {
  if(x=="tea") {
    break
  }
  print(x)
}
```

The next command

```r
for(x in drink) {
  if(x=="lemonade") {
    next
  }
  print(x)
}
```

```r
for(x in drink) {
  if(x=="tea") {
    next
  }
  print(x)
}
```

Loops in R: while loop

```r
i = 1
while(i<10) {
  print(i^2)
  i = i+2
}
```

```r
sumfunction = function() {
  sum = 0
  number = as.integer(readline(prompt="Please select any number less than 25: "))
  while(number <= 25) {
    sum = sum + number
    number = number + 1
  }
  print(paste("The sum of numbers received from the while loop: ", sum))
}
```

```r
sumfunction()
```

The repeat loop

```r
i = 1
repeat {
  print(i^2)
  i = i+2
  if(i>10)
  break
}
```

```r
i = 1
repeat {
  i=i+1
  if(i<10) next
  print(i^2)
  if(i>=13) break
}
```
